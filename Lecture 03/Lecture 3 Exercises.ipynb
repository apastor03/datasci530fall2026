(c) Clean a column

<font size = "3">

- Import the dataset "data_raw/results.csv" to a new object "results"
- Our goal is to replace any string values in the column "milliseconds" <br>
to missing values (NaNs) and covert to numeric
- To do so, go to Lecture 13 and follow the setups under the headings
    - "Extract list of non-numeric values"
    - "Replace certain values"
    - "Convert column to numeric"

HINT: substitute "milliseconds" instead of "alt", <br>
and the dataset "results" instead of "circuits" 


````python
# Write your own code



# Extract list of non-numeric values
non_numeric = results.loc[
    ~pd.to_numeric(results["milliseconds"], errors="coerce").notna(),
    "milliseconds"
].unique()

# Replace non-numeric values with NaN
results["milliseconds"] = results["milliseconds"].replace(non_numeric, np.nan)

# Convert column to numeric
results["milliseconds"] = pd.to_numeric(results["milliseconds"])
````

(g) Recode a numeric colum

<font size = "3">

- Import the "data_raw/races.csv" file
- Recode the "year" column into a new column "year_brackets" <br>
with the following categories

$\qquad$ ``` ["1950-1970","1971-1990","1991-2005", "2006-2020", "2020-onwards"] ```

$\qquad$ HINT: Use the "pd.cut()" command. 

````python
# Write your own code


# Import the dataset
races = pd.read_csv("data_raw/races.csv")

# Define the bins and labels for the year brackets
bins = [1949, 1970, 1990, 2005, 2020, np.inf]
labels = ["1950-1970", "1971-1990", "1991-2005", "2006-2020", "2020-onwards"]

# Recode the year column into new year_brackets
races["year_brackets"] = pd.cut(
    races["year"],
    bins=bins,
    labels=labels,
    include_lowest=True,
    right=False
)

````

(h) Aggregate and query

<font size = "3">

- Import the "data_raw/constructor_standings.csv" file
- Subset the observations for 'raceId == 10' using ```.query()```
- Obtain the sum of "points" by "constructorId" <br> 
(the team) using ``` .groupby().agg() ```
- Subset the teams with a sum of points $\ge 50$
- Store the results of all these operations in a dataframe <br>
```constructor_standings_agg ```

````python
# Write your own code


constructor_standings_agg = (
    constructor_standings
    .query("raceId == 10")
    .groupby("constructorId", as_index=False)["points"]
    .agg("sum")
    .rename(columns={"points": "total_points"})
    .query("total_points >= 50")
)

````

<font size = "5">

(i) Aggregate and sort

<font size = "3">

- Import the "data_raw/constructor_standings.csv" file
- Obtain the sum of "points" by "constructorId" <br> 
(the team) using ``` .groupby().agg() ```
- Sort the teams in descending order using of <br>
the sum of points using ``` .sort_values() ```

````python
# Write your own code



constructor_standings_agg = (
    constructor_standings.groupby("constructorId", as_index=False)["points"]
    .agg("sum")
    .rename(columns={"points": "total_points"})
    .sort_values("total_points", ascending=False)
)

````

(k) Merge dataset

<font size = "3">

- Import the "data_raw/constructor_standings.csv" and <br>
"data_raw/constructors.csv" files
- Create a new dataset with  ```pd.merge()``` using <br>
"constructor_standings" as the primary dataset, and <br>
"constructors" as the secondary dataset, merging on <br>
the column "constructorId"
- To get full points **ONLY** merge the "nationality" column <br>
from the secondary dataset (not all the columns)

HINT: Use ```[[...]]``` to extract a subset of columns <br>
from the secondary dataset before merging, including the <br>
"constructorId" column

````python
# Write your own code

# Import the secondary dataset
constructors = pd.read_csv("data_raw/constructors.csv")

# Merge constructor_standings with only constructorId and nationality from constructors
constructor_standings_merge = pd.merge(
    constructor_standings,
    constructors[["constructorId", "nationality"]],
    on="constructorId",
    how="left"
)




````

(l) Formatted results table

<font size = "3">

-  For the dataset you computed above construct a \
descriptive statistics table
- Round the numbers and change the column names to something \

````python
# Write your own code

summary_table = (
    constructor_standings_merge[["points", "wins_constructor", "positionNumeric"]]
    .describe()
    .round(2)
    .rename(index={
        "points": "Points",
        "wins_constructor": "Wins",
        "positionNumeric": "Position (Numeric)"
    })
    .rename(columns={
        "count": "N",
        "mean": "Mean",
        "std": "Std. Dev.",
        "min": "Min",
        "25%": "25th Percentile",
        "50%": "Median",
        "75%": "75th Percentile",
        "max": "Max"
    })
)

summary_table
````
